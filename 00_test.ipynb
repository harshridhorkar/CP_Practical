```python
# Create a simple if/elif/else to assign grades using new ranges.
marks = float(input('Enter your marks (0-100): '))

if marks < 0 or marks > 100:
    print('Invalid marks. Enter a value between 0 and 100.' )
elif marks >= 85:
    grade = 'A'
elif marks >= 70:
    grade = 'B'
elif marks >= 50:
    grade = 'C'
elif marks >= 40:
    grade = 'D'
else:
    grade = 'F'

print('Your grade is:', grade)
```
