In a batch reactor,reactant A is consumed by a first-order reaction, dC/dt  = -Kc.Use the euler method:C_new = C_old-k*C_old*delta t .Take Co=2.5 mol/L, k= 0.04 min^-1 , delta t = 1min. Write python program using math.exp(),calculate the exact value Coe^(-kt) at 12min and print the absolute error of the euler result.

```python
import math

initial_concentration = 2.5
k = 0.04
delta_t = 1
simulation_time = 12
concentration_euler = initial_concentration

for _ in range(simulation_time):
    concentration_euler = concentration_euler - k * concentration_euler * delta_t

exact_concentration = initial_concentration * math.exp(-k * simulation_time)
absolute_error = abs(exact_concentration - concentration_euler)

print(f"Euler concentration at 12 min: {concentration_euler:.6f} mol/L")
print(f"Exact concentration at 12 min: {exact_concentration:.6f} mol/L")
print(f"Absolute error: {absolute_error:.6f} mol/L")
```

 in a batch reactor,reactant A is consumed by a first-order reaction, dC/dt  = -Kc.Use the euler method:C_new = C_old-k*C_old*delta t .Take Co=2.5 mol/L, k= 0.04 min^-1 , delta t = 1min. Write python program using a while loop, repeat the simulation until conversion reaches 40% or time reaches 100 min .Print the time needed and the final concentration.

```python
initial_concentration = 2.5
k = 0.04
delta_t = 1
concentration = initial_concentration
time = 0
conversion = 0.0

while conversion < 40.0 and time < 100:
    concentration = concentration - k * concentration * delta_t
    time += delta_t
    conversion = ((initial_concentration - concentration) / initial_concentration) * 100

print(f"Time needed: {time} min")
print(f"Final concentration: {concentration:.4f} mol/L")
print(f"Conversion: {conversion:.2f}%")
```

PART B.

Distillation- column sample codes such as DIST2026X9Q3 are stored as text. Write a program that reads code , loops through each character, and prints the number of digits and the sum of the digits.Show a message for an empty code and for a code with no digits .Test with DIST2026X9Q3,an empty code and ABCDEF

```python
def analyze_code(code):
    if code == "":
        print("Message: The code is empty.")
        return

    digit_count = 0
    digit_sum = 0

    for character in code:
        if character.isdigit():
            digit_count += 1
            digit_sum += int(character)

    if digit_count == 0:
        print("Message: The code contains no digits.")
    else:
        print("Code:", code)
        print("Number of digits:", digit_count)
        print("Sum of digits:", digit_sum)
```

```python
analyze_code("DIST2026X9Q3")
print()
analyze_code("")
print()
analyze_code("ABCDEF")
```

Print the number of possible mixtures for n=9,r=3,and show the message for n=6,r=8.PART A.a laboratory has n candidate solvents and wants to test r of them together in a mixed solvent(order does not matter).Write a recursive function factorial(n)(remember the base case for 0factori

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)


def combination(n, r):
    if n < 0 or r < 0:
        print("Error: n and r must be non-negative integers.")
        return None
    if r > n:
        print("Error: r cannot be greater than n.")
        return None

    return factorial(n) // (factorial(r) * factorial(n - r))
```

```python
mixtures = combination(9, 3)
print("Number of possible mixtures for n=9, r=3:", mixtures)

combination(6, 8)
```

PART B.During a crystallisation run, samples are taken at times=Fibonacci number * 4min.Write an iterative function fib_iter(n) that returns a list of the first n Fibonacci numbers, and a recursive function fib_rec(k) that returns the k-th Fibonacci number (Fo=0,F1=1). For n=10,check that both functions agree and print a table:Sample No. / Fibonacci tag / Sampling time (min).

```python
def fib_iter(n):
    if n <= 0:
        return []

    fibonacci = [0]
    if n == 1:
        return fibonacci

    fibonacci.append(1)
    for index in range(2, n):
        fibonacci.append(fibonacci[index - 1] + fibonacci[index - 2])

    return fibonacci
```

```python
def fib_rec(k):
    if k == 0:
        return 0
    if k == 1:
        return 1
    return fib_rec(k - 1) + fib_rec(k - 2)
```

```python
n = 10
iterative_values = fib_iter(n)
recursive_values = [fib_rec(k) for k in range(n)]

print("Both functions agree:", iterative_values == recursive_values)
print(f"{'Sample No.':>10} {'Fibonacci tag':>15} {'Sampling time (min)':>20}")

for sample_number, fibonacci_tag in enumerate(iterative_values, start=1):
    sampling_time = fibonacci_tag * 4
    print(f"{sample_number:>10} {fibonacci_tag:>15} {sampling_time:>20}")
```

Ethylene gas is compressed slowly in a piston-cylinder at constant temperature.
volume_L        1        2        3         4        5 
Pressure_bar    5.03     2.52     1.65     1.26      1.01
plot pressure vs. volume as a scatter plot.Calculate k as the mean of P*V AND draw the reference curve P = k/V on the same graph. Add title,labels with units and legend.

```python
import numpy as np
import matplotlib.pyplot as plt

volume_l = np.array([1, 2, 3, 4, 5])
pressure_bar = np.array([5.03, 2.52, 1.65, 1.26, 1.01])
k = np.mean(pressure_bar * volume_l)

volume_curve = np.linspace(volume_l.min(), volume_l.max(), 200)
pressure_curve = k / volume_curve

plt.figure(figsize=(8, 5))
plt.scatter(volume_l, pressure_bar, color="tab:blue", label="Measured pressure")
plt.plot(volume_curve, pressure_curve, color="tab:red", label=f"Reference curve: P = {k:.2f}/V")
plt.title("Pressure-Volume Relationship for Ethylene Gas")
plt.xlabel("Volume (L)")
plt.ylabel("Pressure (bar)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

print(f"k = mean(P * V) = {k:.4f} bar.L")
```
